# Business Requirements

KPI'S Requirements

```python
# Total Sales
total_sales = df["Sales"].sum()

# Average Sales
avg_sales = df["Sales"].mean()

# no. of Item Sold 
no_of_item_sold = df["Sales"].count()

# Average Rating 
avg_rating =df["Rating"].mean()

# Disply

print(f"Total Sales : ${total_sales :,.0f}")
print(f"Average Sales : ${avg_sales : ,.0f}")
print(f"Numbers of Item Sold : {no_of_item_sold:,.0f}")
print(f"Average Rating :{avg_rating:,.0f}")
```

# Chart Requirement 

Total sales by fat contant

```python
sales_by_fat = df.groupby("Item Fat Content")["Sales"].sum()
sales_by_fat
plt.pie(sales_by_fat,labels = sales_by_fat.index,
        autopct = '%.1f%%',
        startangle=90)
plt.title("Sales by Fat Contant",color="red",fontsize=16)
plt.axis('equal')
plt.show()
```

Total Sales by Item Type 

```python
sales_by_type = df.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)
sales_by_type

plt.figure(figsize=(10,6))
bars = plt.bar(sales_by_type.index,sales_by_type.values)

plt.xticks(rotation=-90)
plt.xlabel("Itme Type")
plt.ylabel("Total Sales")
plt.title("Total Sales By Item Type",color='black',fontsize='16')

for bar in bars :
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height(),
             f"{bar.get_height():,.0f}", ha = 'center',va = 'bottom' , fontsize = 8)
plt.tight_layout()
plt.show()
             
```

Fat Content by Outlet for Total Sales

```python
group = df.groupby(['Outlet Location Type','Item Fat Content'])['Sales'].sum().unstack()
group = group[['Regular','Low Fat']]

ax = group.plot(kind='bar',figsize=(8,5),
                title= 'Outlet Tier by Item Fat Content')

plt.xlabel('Outlet Location Tier')
plt.ylabel('Total Sales')
plt.legend(title='Item Fat Content')
plt.tight_layout()
plt.show()
```

Total Sales By outlet Estabilishemt 

```python
sales_by_year = df.groupby('Outlet Establishment Year')['Sales'].sum().sort_index()

plt.figure(figsize=(9,5))

plt.plot(sales_by_year.index,
         sales_by_year.values,marker='o',
         linestyle='-')

plt.xlabel('Outlet Establishment Year')
plt.ylabel('Total Sales')
plt.title("Outlet Establishment")

for x,y in zip(sales_by_year.index,sales_by_year.values):
    plt.text(x,y,f'{y:,.0f}',ha = 'center', va = 'bottom', fontsize = 8)

plt.tight_layout()
plt.show()
```

Sales by Outlet Size

```python
sales_by_size = df.groupby('Outlet Size')['Sales'].sum()

plt.figure(figsize=(4,4))
plt.pie(sales_by_size, labels =sales_by_size.index, autopct='%1.1f%%', startangle=90)
plt.title('Outlet Size')
plt.tight_layout()
plt.show()
```

Sales by Outlet Location Type

```python
sales_by_location = df.groupby('Outlet Location Type')['Sales'].sum().reset_index()
sales_by_location = sales_by_location.sort_values('Sales',ascending=False)

plt.figure(figsize=(8,3))
ax = sns.barplot(x='Sales', y='Outlet Location Type', data=sales_by_location)

plt.title('Total Sales by Outlet Location type')
plt.xlabel('Total Sales')
plt.ylabel('Outlet Location Type')
plt.tight_layout()
plt.show()
```
